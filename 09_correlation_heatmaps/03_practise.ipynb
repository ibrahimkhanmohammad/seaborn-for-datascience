## Dataset: Student Performance

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme()
sns.color_palette()

data = {
    "Student": [
        "Ali",
        "Ahmed",
        "Sara",
        "Zoya",
        "Omar",
        "Aisha",
        "Usman",
        "Fatima",
        "Hassan",
        "Maryam",
        "Bilal",
        "Noor",
    ],
    "Hours_Studied": [1, 2, 3, 4, 5, 6, 7, 8, 2, 5, 6, 9],
    "Attendance": [50, 55, 60, 65, 70, 75, 80, 90, 58, 72, 85, 95],
    "Marks": [30, 35, 45, 50, 60, 68, 75, 90, 38, 65, 78, 95],
    "Sleep_Hours": [9, 8, 8, 7, 7, 6, 6, 5, 9, 7, 6, 5],
    "Assignments_Completed": [2, 3, 4, 5, 6, 7, 8, 10, 3, 7, 8, 11],
}

df = pd.DataFrame(data)
```

## Task 1 — Correlation Matrix
- Calculate the correlation matrix.
- Display it using print().
- Make sure non-numeric columns are excluded.

```python
corr = df.corr(numeric_only=True)

print(corr)
```

## Task 2 — Basic Heatmap

Create a heatmap with:
- Figure size: (10, 7)
- Annotations enabled.
- Colour map: coolwarm.
- Appropriate title.
- Use Matplotlib's object-oriented approach.

```python
fig, ax = plt.subplots(figsize=(10, 7))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    ax=ax,
)
ax.set_title("Basic Heatmap")
plt.show()
```

## Task 3 — Formatting

Create another heatmap with:
- Values displayed to 2 decimal places.
- vmin=-1 and vmax=1.
- linewidths=0.5.
- Colour bar enabled.

```python
fig, ax = plt.subplots(figsize=(10, 7))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    ax=ax,
    cbar=True,
    linewidths=0.5,
    fmt=".2f",
)
ax.set_title("Basic Heatmap")
plt.show()
```

## Task 4 — Customisation

Create a heatmap with:
- Square-shaped cells.
- Colour map: RdYlBu_r.
- Annotations enabled.
- Remove the colour bar.

```python
fig, ax = plt.subplots(figsize=(10, 7))

sns.heatmap(
    corr,
    annot=True,
    cmap="RdYlBu_r",
    ax=ax,
    square=True,
    cbar=False,
    linewidths=0.1,
    fmt=".2f",
)
ax.set_title("Basic Heatmap")
plt.show()
```
