## 1. What is correlation?
Correlation tells us whether two numerical variables are related, and how strongly they move together.

Imagine we're analysing students:
- How do study hours relate to exam marks?
- How does attendance relate to exam marks?
- How does sleep relate to study hours?
For example, if students who study more generally score higher marks, then Study Hours and Marks have a positive correlation.

## 2. Three types of correlation
Correlation values range from -1 to +1.
Positive correlation -> 0 to +1

When one variable increases, the other tends to increase too.

Negative correlation
-1 to 0

When one variable increases, the other tends to decrease.

Zero or near-zero correlation
Around 0

There is little or no linear relationship between the variables.

Important: A correlation near zero does not prove there is no relationship at all. There could be a nonlinear relationship that Pearson correlation doesn't capture.

## 3. Understanding correlation values
    Correlation	Interpretation
    +1.00	    Perfect positive
    +0.85	    Strong positive
    +0.50	    Moderate positive
    +0.10	    Weak positive
    0.00	    No linear correlation
    -0.50	    Moderate negative
    -0.90	    Strong negative
    -1.00	    Perfect negative

These strength descriptions are general guidelines, not strict universal rules.

## 4. What is Pearson correlation?
Pandas' .corr() calculates Pearson correlation by default.

Pearson correlation measures the strength and direction of a linear relationship between two numerical variables.

For example:
- Study hours increase.
- Marks also tend to increase.
- Pearson correlation might be 0.90.
That means a strong positive linear relationship.

## 5. Create a correlation matrix using Pandas
Let's use a student-performance dataset.

We have four numerical columns:
- Hours_Studied
- Attendance
- Marks
- Sleep_Hours

What does df.corr() do?

It calculates the correlation between every pair of numerical columns and returns a correlation matrix.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "Hours_Studied": [1, 2, 3, 4, 5, 6, 7, 8],
    "Attendance": [50, 55, 60, 65, 70, 75, 80, 85],
    "Marks": [35, 40, 50, 55, 65, 70, 80, 90],
    "Sleep_Hours": [8, 7, 8, 6, 7, 6, 5, 5]
}

df = pd.DataFrame(data)

print(df)

corr = df.corr()

print(corr)
```

Notice three things:
1. The diagonal values are always 1.00, because every variable is perfectly correlated with itself.
2. The matrix is symmetrical: the correlation between A and B is the same as B and A.
3. The values range from -1 to +1.

If your DataFrame also contains text columns, Pandas normally excludes non-numeric columns from the calculation. You can explicitly write:

`corr = df.corr(numeric_only=True)`
