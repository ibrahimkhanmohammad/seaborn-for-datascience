Create a histogram with:
1. X-axis: Marks
2. Color: skyblue
3. Bins: 5
4. Figure size: (10, 6)
5. Title: Distribution of Student Marks
6. X-axis label: Marks
7. Y-axis label: Frequency

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    "Student": ["Ali", "Ahmed", "Sara", "Zoya", "Omar",
                "Aisha", "Usman", "Fatima", "Hassan", "Maryam"],
    "Marks": [35, 40, 45, 50, 55, 60, 65, 70, 80, 90]
}

df = pd.DataFrame(data)
````

````python
fig, ax = plt.subplots(figsize=(10, 6))

sns.histplot(
    data=df,
    x="Marks",
    ax=ax,
    color="skyblue",
    bins=5,
)
ax.set_title("Distribution of Student Marks")
ax.set_xlabel("Marks")
ax.set_ylabel("Frequency")

plt.show()
````

````python
fig, ax = plt.subplots(figsize=(10, 6))

sns.histplot(
    data=df,
    x="Marks",
    bins=5,
    kde=True,
    color="skyblue",
    ax=ax,
)

ax.set_title("Distribution of Student Marks")
ax.set_xlabel("Marks")
ax.set_ylabel("Frequency")

plt.show()
````
