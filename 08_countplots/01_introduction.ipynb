## 1. What is a count plot?
A count plot is used to visualize the frequency (number of occurrences) of each category in a dataset.

In simple words, it answers:

How many times does each category appear?

Imagine we have this student dataset:           
```console
    Student	Department      
    Ali	        CSE
    Ahmed	ECE
    Sara	CSE
    Zoya	IT
    Omar	CSE
    Aisha	ECE
```

We want to know how many students belong to each department.

The result:     
```console
    Department	Count
    CSE     	3
    ECE     	2
    IT      	1
```

A count plot automatically calculates these frequencies and visualizes them.

## 2. Why do Data Scientists use count plots?
In EDA, we frequently work with categorical columns.

For example:
- Gender: Male, Female.
- Department: CSE, ECE, IT.
- Payment method: Cash, UPI, Card.
- Customer type: New, Returning.
- Product category: Electronics, Clothing, Furniture.
We often need to understand how observations are distributed across these categories.

For example, imagine a customer dataset contains 10,000 records.

We want to know:
- How many customers use UPI?
- How many customers use cash?
- How many customers use cards?
A count plot makes this easy.

## 3. Syntax — sns.countplot()
Let's create a practical dataset.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    "Department": [
        "CSE", "CSE", "ECE", "IT",
        "CSE", "ECE", "IT", "CSE",
        "ECE", "CSE", "IT", "ECE"
    ]
}

df = pd.DataFrame(data)
```

Now visualize the frequency of each department:

````python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme()
sns.color_palette()

data = {
    "Department": [
        "CSE", "CSE", "ECE", "IT",
        "CSE", "ECE", "IT", "CSE",
        "ECE", "CSE", "IT", "ECE"
    ]
}

df = pd.DataFrame(data)

fig, ax = plt.subplots(figsize=(8, 5))

sns.countplot(
    data=df,
    x="Department",
    ax=ax
)

ax.set_title("Number of Students by Department")
ax.set_xlabel("Department")
ax.set_ylabel("Number of Students")

plt.show()


# The plot automatically counts the occurrences.

# Expected output:
# - CSE → 5
# - ECE → 4
# - IT → 3
````

## 4. Understanding x and y
You can create a count plot in two orientations.

A. Vertical count plot      
`sns.countplot(    data=df,    x="Department",    ax=ax)
`

- X-axis → Categories.  
- Y-axis → Frequency.   

B. Horizontal count plot        
`sns.countplot(    data=df,    y="Department",    ax=ax)
`

- Y-axis → Categories.
- X-axis → Frequency.

Use horizontal orientation when category names are long.

## 5. Important parameters
Let's understand the parameters you'll commonly use.        
```console
Parameter	Purpose
data	    DataFrame containing data
x	    Categorical variable for horizontal axis
y	    Categorical variable for vertical axis
hue	    Adds another categorical grouping
order	    Controls category display order
palette	    Assigns colors to categories
color	    Uses a single color
stat	    Controls what the bar height represents
ax          Matplotlib Axes object
```

### A. hue — Grouping by another category
Suppose we have student gender as an additional column.

````python
data = {
    "Department": ["CSE", "CSE", "CSE", "ECE", "ECE", "ECE", "IT", "IT"],
    "Gender": ["Male", "Female", "Male", "Female", "Male", "Female", "Male", "Male"],
}

df = pd.DataFrame(data)
fig, ax = plt.subplots(figsize=(10, 6))
sns.countplot(data=df, x="Department", hue="Gender", ax=ax)

plt.show()
````

What happens?

Each department gets separate bars for Male and Female.

This helps us compare gender frequencies across departments.

### B. order — Changing category order
By default, Seaborn generally uses the order in which categories appear in the data (unless categorical ordering is defined).

You can control the order:

````python
fig, ax = plt.subplots(figsize=(10, 6))

sns.countplot(
    data=df,
    x="Department",
    order=["IT", "ECE", "CSE"],
    ax=ax
)

plt.show()
````

Now the categories appear in this order:

IT → ECE → CSE.

This is useful when you want a consistent presentation.

### C. stat — Count vs proportion
This is an important parameter.

By default: 

`stat="count"
`
It displays the number of observations.

You can also use:       
`stat="percent"
`

It displays each category's percentage of the total observations.

For our original dataset:
```console
Department	Count	Percentage
CSE     	5	41.67%
ECE	        4	33.33%
IT	        3	25%
```

Example:    

```python
sns.countplot(
    data=df,
    x="Department",
    stat="percent",
    ax=ax
)
```

Note: stat="percent" is available in modern Seaborn versions. If your installed version doesn't support it, check your Seaborn version before troubleshooting.

## 6. Bar plot vs Count plot — Final comparison
    This is important enough to remember throughout your Data Science journey.
    Feature	                sns.barplot()	                sns.countplot()
    Purpose	                Compare numerical values	Count categories
    Requires numerical y	Usually yes	                No
    Default calculation	Mean	                        Count
    Aggregation	        Mean, sum, median, etc.	        Frequency
    Typical question	How much?	                How many?

## Example:

Imagine we have a product dataset.
- barplot: What is the average sales value for each product?
- countplot: How many times does each product appear in the dataset?
That's the difference.

## 7. Real-world EDA applications
Count plots are especially useful for:
1. Customer analysis: Number of customers by membership type.
2. Banking: Number of customers by account type.
3. E-commerce: Number of orders by payment method.
4. Education: Number of students by department.
5. Machine learning: Checking class distribution in classification datasets.
For ML, checking class imbalance is particularly important.

For example, if a fraud detection dataset contains 9,900 normal transactions and only 100 fraudulent transactions, a count plot can quickly reveal that imbalance.
