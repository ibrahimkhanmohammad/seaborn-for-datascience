## Task 1 — Basic count plot
Create a count plot showing the number of orders for each product.

````python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme()
sns.color_palette()


data = {
    "Product": [
        "Laptop", "Mobile", "Laptop", "Tablet",
        "Mobile", "Laptop", "Tablet", "Mobile",
        "Laptop", "Mobile", "Tablet", "Laptop",
        "Mobile", "Tablet", "Laptop"
    ],
    "Region": [
        "South", "North", "East", "South",
        "East", "North", "West", "South",
        "East", "North", "West", "South",
        "East", "North", "West"
    ],
    "Payment": [
        "UPI", "Card", "Cash", "UPI",
        "UPI", "Card", "Cash", "UPI",
        "Cash", "Card", "UPI", "UPI",
        "Cash", "Card", "UPI"
    ]
}

df = pd.DataFrame(data)
````

````python
fig, ax = plt.subplots(figsize=(10, 6))

sns.countplot(
    data=df,
    x="Product",
    ax=ax,
    color="skyblue",
    edgecolor="black",
)

ax.set_title("No of orders per Product")
ax.set_xlabel("Product")
ax.set_ylabel("Frequency")
plt.show()
````

## Task 2 — Hue
Create a count plot showing product counts separated by region.

Use hue="Region".

````python
fig, ax = plt.subplots(figsize=(10, 6))

sns.countplot(
    data=df,
    x="Product",
    ax=ax,
    hue="Region",
)

ax.set_title("No of product orders per Region")
ax.set_ylabel("Frequency")
fig.tight_layout()
plt.show()
````

## Task 3 — Horizontal count plot
Display the number of orders for each payment method horizontally.

````python
fig, ax = plt.subplots(figsize=(10, 6))

sns.countplot(
    data=df,
    y="Payment",
    ax=ax,
)

ax.set_title("No of Payments")
ax.set_xlabel("Frequency")
plt.show()
````

## Task 4 — Order
Display products in this exact order:
1. Tablet
2. Mobile
3. Laptop

````python
fig, ax = plt.subplots(figsize=(10, 6))

sns.countplot(
    data=df,
    x="Product",
    ax=ax,
    order=["Tablet", "Mobile", "Laptop"]
)

ax.set_title("No of orders per Product")
ax.set_ylabel("Frequency")
plt.show()
````

## Task 5 — Percentage
Create a count plot showing the percentage of orders for each payment method using stat="percent".

````python
fig, ax = plt.subplots(figsize=(10, 6))

sns.countplot(
    data=df,
    x="Payment",
    ax=ax,
    stat="percent",
)

ax.set_title("Percentage of Payment methods across orders")
ax.set_ylabel("Percentage (%)")
plt.show()
````

## Conceptual Questions

1. Highest product orders → Laptop. 
2. Most frequently used payment method → UPI. 
3. Purpose of hue → Compare regional orders.
4. Counting orders vs sales revenue → Frequency vs sum of revenue.

- Counting orders: How many transactions occurred?
- Sales revenue: How much money was generated?      
For example, 5 orders worth ₹1,000 each produce ₹5,000 revenue.

Both can be aggregations, but they measure different things.
