## Lesson — sns.kdeplot()
KDE stands for Kernel Density Estimation.

Its purpose is to estimate and visualize the distribution of numerical data using a smooth curve.

For example:

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    "Student": ["Ali", "Ahmed", "Sara", "Zoya", "Omar",
                "Aisha", "Usman", "Fatima", "Hassan", "Maryam"],
    "Marks": [35, 40, 45, 50, 55, 60, 65, 70, 80, 90]
}

df = pd.DataFrame(data)

fig, ax = plt.subplots(figsize=(10, 6))

sns.kdeplot(
    data=df,
    x="Marks",
    color="green",
    fill=True,
    ax=ax
)

ax.set_title("Distribution of Student Marks")
ax.set_xlabel("Marks")
ax.set_ylabel("Density")

plt.show()
```

## Understand these parameters
    Parameter	    Purpose
    data=df	            DataFrame
    x="Marks"	    Numerical column
    color="green"	    Curve color
    fill=True	    Fills the area under the curve
    ax=ax	            Matplotlib plotting area

## Histogram vs KDE
    Histogram	                        KDE
    Uses bars	                        Uses smooth curve
    Shows frequency/count by default	Shows estimated density
    Divides data into bins	                Estimates a continuous distribution
    Sensitive to bin selection	        Sensitive to bandwidth

Important: KDE bandwidth controls how smooth the curve is. Too much smoothing can hide details; too little can make the curve irregular.
