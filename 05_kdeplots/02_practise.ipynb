## Use your existing df and create a standalone KDE plot with:
1. sns.kdeplot()
2. X-axis: Marks
3. Color: purple
4. fill=True
5. Figure size: (10, 6)
6. Appropriate title and axis labels.

````python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    "Student": ["Ali", "Ahmed", "Sara", "Zoya", "Omar",
                "Aisha", "Usman", "Fatima", "Hassan", "Maryam"],
    "Marks": [35, 40, 45, 50, 55, 60, 65, 70, 80, 90]
}

df = pd.DataFrame(data)

fig, ax = plt.subplots(figsize=(10, 6))

sns.kdeplot(
    data=df,
    x="Marks",
    color="purple",
    fill=True,
    ax=ax,
)

ax.set_title("Distribution of Student Marks")
ax.set_xlabel("Marks")
ax.set_ylabel("Density")

plt.show()
````

## Conceptual question

The difference is:
- Histogram counts how many observations fall into each bin.
- KDE estimates the underlying distribution of the data and represents it as a smooth density curve.
For example, imagine 100 students' marks.
Histogram - Shows how many observations fall within each interval.

KDE - Estimates where observations are concentrated, using a smooth curve.

One important fact: KDE's vertical axis is density, not frequency. The total area under a properly normalized KDE curve is 1.       
Remember this       
```console
    Function	                What it tells us
    sns.histplot()	        How many observations are in each bin?
    sns.kdeplot()	        How is the data distributed?
    sns.histplot(kde=True)	Both frequency and estimated density
```
