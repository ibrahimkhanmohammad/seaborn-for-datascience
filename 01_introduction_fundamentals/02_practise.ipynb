## Task 1 — Load the dataset
Use sns.load_dataset() to load iris.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = sns.load_dataset("iris")
```

## Task 2 — Explore the dataset
- Display the first 10 rows.
- Check dataset information.
- Check statistical description.
- Check missing values.

```python
df.head(10)
df.info()
df.describe()
df.isna().sum()
```

## Task 3 — Create a scatterplot
Requirements:
- X-axis: sepal_length
- Y-axis: petal_length
- Figure size: (10, 6)
- Color: Green
- Title: Sepal Length vs Petal Length
- Use fig, ax = plt.subplots().

```python
fig, ax = plt.subplots(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="sepal_length",
    y="petal_length",
    color="green",
    ax=ax,
)

ax.set_title("Sepal Length vs Petal Length")
ax.set_xlabel("Sepal Length")
ax.set_ylabel("Petal Length")

plt.show()
```

## Task 4 — Answer these questions

1. What is Seaborn?             
Seaborn is a Python data visualization library used to create attractive statistical graphics. It is built on Matplotlib.

2. Why do we use Seaborn in Data Science?           
Seaborn makes it easier to visualize statistical relationships, distributions, and patterns in datasets during Exploratory Data Analysis (EDA).

3. What does data=df mean?          
It tells Seaborn to use the DataFrame stored in the variable df as the source of data for the plot.

4. What is the purpose of ax=ax?                       
It tells Seaborn to draw its visualization on the Matplotlib Axes object stored in the variable ax.

5. Difference between Matplotlib and Seaborn?           
- Matplotlib: General-purpose visualization library offering detailed control over plots.
- Seaborn: Built on Matplotlib and provides convenient functions for statistical visualization with attractive default styles.
