## 1. What is Seaborn?
Seaborn is a Python data visualization library used to create statistical graphs and understand patterns in datasets.

In Data Science, we mainly use it during Exploratory Data Analysis (EDA).

For example, imagine you have a dataset containing:
- Students' study hours
- Students' marks
- Students' attendance
You want to understand:
- Does studying more lead to higher marks?
- How are marks distributed?
- Are there any unusual values?
- Which variables are related?
Seaborn helps us answer these questions visually.

## 2. Seaborn vs Matplotlib

Matplotlib - Detailed control over figures, axes and visual elements.

Seaborn - Convenient statistical graphs with attractive default styling.


        Matplotlib	                    Seaborn
        General visualization library   Statistical visualization library
        More manual customization	    Convenient defaults
        Often uses X and Y arrays	    Works directly with DataFrames
        ax.scatter()	            sns.scatterplot()

Important: Seaborn is built on top of Matplotlib. They work together, not against each other.

## 3. Importing Seaborn and Set the theme
This applies Seaborn's default plotting style.

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme()
````

## 4. Understanding Seaborn datasets
Seaborn provides built-in example datasets that we can use for practice.

Let's load the famous tips dataset.

The dataset contains information about restaurant bills and tips.

Let's inspect it using your Pandas knowledge.

````python
df = sns.load_dataset("tips")

df.head()
df.describe()
df.info()
df.isnull().sum()
````

## 5. Your first Seaborn graph
Let's understand the most important plotting structure.

````python
fig, ax = plt.subplots(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="total_bill",
    y="tip",
    ax=ax,
    color="green"
)

ax.set_title("Total Bill vs Tip")
ax.set_xlabel("Total Bill")
ax.set_ylabel("Tip")

plt.show()
````

Understand every parameter
   ```console
    Parameter	            Explanation
    data=df	    DataFrame containing the data
    x="total_bill"	    Column plotted on X-axis
    y="tip"	    Column plotted on Y-axis
    ax=ax	            Matplotlib axes where graph is drawn
    color="green"	    Color of points
   ```

Remember this structure:
```python
fig, ax = plt.subplots()

sns.plot_function(
    data=df,
    x="column1",
    y="column2",
    ax=ax
)

plt.show()
```
We'll reuse this structure for most of our Seaborn plots.
