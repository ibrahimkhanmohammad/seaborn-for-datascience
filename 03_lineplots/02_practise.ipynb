## Task 1 Create a line plot for monthly expenses.

Requirements:
1. Use the same DataFrame.
2. X-axis: Month.
3. Y-axis: Expenses.
4. Use a red line.
5. Use circular markers.
6. Figure size: (10, 6).
7. Add title: Monthly Expenses Trend.
8. Add appropriate X-axis and Y-axis labels.
9. Use fig, ax = plt.subplots() and sns.lineplot().

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Sales": [12000, 15000, 18000, 14000, 22000, 25000],
    "Expenses": [8000, 9000, 11000, 10000, 13000, 15000]
}

df = pd.DataFrame(data)

fig,ax=plt.subplots(figsize=(10,6))
sns.lineplot(data=df,x="Month",y="Expenses",ax=ax,marker="o",color="red",)

ax.set_title("Monthly Expenses Trend")
ax.set_xlabel("Month")
ax.set_ylabel("Expenses")

plt.show()
````

## One conceptual question:
Why would we use a line plot instead of a scatter plot to visualize monthly sales trends?

- Scatter plots are useful for examining relationships between two numerical variables.
- Line plots are useful for understanding trends and changes across an ordered variable, such as months.

A line plot connects data points, making the direction of change easier to follow.

For example:
- January → February: Expenses increased.
- March → April: Expenses decreased.
- April → June: Expenses increased.
That's exactly what we want to understand in trend analysis.

Remember these two plots
```console
    Plot	            Main purpose
    sns.scatterplot()	    Relationships between variables
    sns.lineplot()	    Trends and changes
```
