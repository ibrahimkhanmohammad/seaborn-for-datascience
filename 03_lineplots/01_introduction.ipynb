## 1. What is a line plot?
A line plot is used to visualize trends and changes in numerical values, usually over time.

For example:
- Monthly sales
- Daily temperature
- Yearly revenue
- Website visitors over time

## 2. Matplotlib vs Seaborn
Matplotlib:         
`fig, ax = plt.subplots(figsize=(10, 6))ax.plot(months, sales)plt.show()
`       
    
Seaborn:            
`fig, ax = plt.subplots(figsize=(10, 6))sns.lineplot(    data=df,    x="Month",    y="Sales",    ax=ax)plt.show()
`

Notice: We use sns.lineplot() instead of ax.plot().

Now visualize monthly sales:

````python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Sales": [12000, 15000, 18000, 14000, 22000, 25000],
    "Expenses": [8000, 9000, 11000, 10000, 13000, 15000]
}

df = pd.DataFrame(data)


fig, ax = plt.subplots(figsize=(10, 6))

sns.lineplot(
    data=df,
    x="Month",
    y="Sales",
    marker="o",
    color="green",
    ax=ax
)

ax.set_title("Monthly Sales Trend")
ax.set_xlabel("Month")
ax.set_ylabel("Sales")

plt.show()
````
