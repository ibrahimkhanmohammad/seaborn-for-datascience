## Seaborn Scatterplot in detail
Now we'll learn one important feature that makes Seaborn especially useful for EDA: hue.

In Matplotlib, you might manually assign different colors to different groups. In Seaborn, you can do this using one parameter.

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

df = sns.load_dataset("iris")

fig, ax = plt.subplots(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="sepal_length",
    y="petal_length",
    hue="species",          # Here, hue="species" automatically assigns different colors to the three iris species and creates a legend.
    ax=ax
)

ax.set_title("Sepal Length vs Petal Length by Species")
ax.set_xlabel("Sepal Length")
ax.set_ylabel("Petal Length")

plt.show()
```

## Next: style and size (quick introduction)
Seaborn can distinguish groups not only through colors but also through marker shapes and sizes.
- hue differentiates species using colors.
- style differentiates species using marker shapes.

```python
fig, ax = plt.subplots(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="sepal_length",
    y="petal_length",
    hue="species",
    style="species",
    ax=ax
)

ax.set_title("Iris Species Comparison")
plt.show()
```
