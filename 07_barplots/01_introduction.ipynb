## 1. What is a bar plot?
A bar plot is used to compare a numerical value across different categories.

For example, imagine we have student marks:
```console
    Student	Marks
    Ali 	85
    Ahmed	72
    Sara	90
    Zoya	65
    Omar	78
```

We can use a bar plot to compare the marks of each student.

In Data Science and EDA, bar plots are commonly used to compare:
- Average salary across departments.
- Average marks across classes.
- Total sales across products.
- Average customer spending across segments.

## 2. Important concept: Bar plot vs Count plot
This is something you must understand.

Both look similar, but they answer different questions.

Bar plot (sns.barplot())

Displays a statistical estimate of a numerical variable for each category.

Example:
```console
        Department	Salary
        CSE	        25000
        CSE	        30000
        CSE	        35000
        ECE	        20000
        ECE	        25000
        ECE	        30000
```

A bar plot can calculate the average salary for each department.

Count plot (sns.countplot())

Displays how many observations belong to each category.

```console
Example:
        Department	Number of employees
        CSE	        3
        ECE	        3
```

Remember:
- barplot → How much?
- countplot → How many?

## 3. Important concept: Aggregation
This is the most important part of sns.barplot().

By default, Seaborn calculates the mean (average) of the numerical variable for each category.

Consider this dataset:

````python
import pandas as pd

data = {
    "Department": ["CSE", "CSE", "CSE", "ECE", "ECE", "ECE"],
    "Salary": [25000, 30000, 35000, 20000, 25000, 30000]
}

df = pd.DataFrame(data)
````

Let's calculate manually.

CSE average:

${25000+30000+35000}/{3}=30000$

ECE average:

${20000+25000+30000}/{3}=25000$

Therefore, the bar plot displays:               
   
```console
Department	Average salary
CSE	        ₹30,000
ECE	        ₹25,000
```

Notice something important: the bar plot doesn't necessarily display every employee's salary. It summarizes the values using an aggregation function.
 

## 4. Syntax — sns.barplot()

````python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    "Department": ["CSE", "CSE", "CSE", "ECE", "ECE", "ECE"],
    "Salary": [25000, 30000, 35000, 20000, 25000, 30000]
}

df = pd.DataFrame(data)

fig, ax = plt.subplots(figsize=(8, 5))
sns.set_theme()
sns.color_palette()
sns.barplot(
    data=df,
    x="Department",
    y="Salary",
    color="skyblue",
    ax=ax
)

ax.set_title("Average Salary by Department")
ax.set_xlabel("Department")
ax.set_ylabel("Average Salary")

plt.show()
````

## A. estimator
By default:
`sns.barplot(data=df, x="Department", y="Salary", estimator="mean")
`


This calculates the average salary.

You can also use:           
`estimator="sum"`

To calculate the total salary for each department.
`estimator="median"
`

To calculate the median salary.

For example, if CSE salaries are 25,000, 30,000 and 35,000:
- Mean = 30,000
- Sum = 90,000
- Median = 30,000

## B. errorbar
You may notice small black lines extending above the bars - These are called error bars.

By default, Seaborn's bar plot generally displays a 95% confidence interval for the estimated mean.

They communicate uncertainty around the estimate.

You can remove them:
`sns.barplot(    data=df,    x="Department",    y="Salary",    errorbar=None,    ax=ax)`

Don't confuse error bars with actual data errors or outliers.

## 7. When do Data Scientists use bar plots?
- Comparing average salaries across departments.
- Comparing revenue across product categories.
- Comparing average student marks across classes.
- Comparing average customer spending across segments.
- Comparing conversion rates across marketing channels.
Important EDA tip: A bar plot can hide the variation between individual observations. If you need to understand the full distribution, use a box plot or violin plot alongside it.
