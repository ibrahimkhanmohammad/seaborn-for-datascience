## Task 1: Create a bar plot showing average sales for each product.

````python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    "Product": [
        "Laptop",
        "Laptop",
        "Laptop",
        "Mobile",
        "Mobile",
        "Mobile",
        "Tablet",
        "Tablet",
        "Tablet",
    ],
    "Sales": [50000, 60000, 70000, 20000, 25000, 30000, 15000, 18000, 22000],
    "Region": [
        "South",
        "North",
        "East",
        "South",
        "North",
        "East",
        "South",
        "North",
        "East",
    ],
}

df = pd.DataFrame(data)

fig, ax = plt.subplots(figsize=(10, 6))
sns.set_theme()
sns.color_palette()

sns.barplot(
    data=df,
    x="Product",
    y="Sales",
    errorbar=None,
    ax=ax,
)
ax.set_title("Average Sales by Product")
ax.set_xlabel("Product")
ax.set_ylabel("Average Sales")
plt.show()
````

## Task 2: Create a bar plot showing total sales for each product using estimator.

````python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    "Product": [
        "Laptop",
        "Laptop",
        "Laptop",
        "Mobile",
        "Mobile",
        "Mobile",
        "Tablet",
        "Tablet",
        "Tablet",
    ],
    "Sales": [50000, 60000, 70000, 20000, 25000, 30000, 15000, 18000, 22000],
    "Region": [
        "South",
        "North",
        "East",
        "South",
        "North",
        "East",
        "South",
        "North",
        "East",
    ],
}

df = pd.DataFrame(data)

fig, ax = plt.subplots(figsize=(10, 6))
sns.set_theme()
sns.color_palette()

sns.barplot(
    data=df,
    x="Product",
    y="Sales",
    errorbar=None,
    ax=ax,
    estimator=sum,
)

plt.show()
````
