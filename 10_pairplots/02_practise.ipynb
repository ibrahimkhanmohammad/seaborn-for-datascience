## Dataset: Student Performance
We'll use this dataset for all tasks.

````python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    "Student": ["Ali", "Ahmed", "Sara", "Zoya", "Omar",
                "Aisha", "Usman", "Fatima", "Hassan", "Maryam",
                "Bilal", "Noor"],

    "Hours_Studied": [1, 2, 3, 4, 5, 6, 7, 8, 2, 5, 6, 9],

    "Attendance": [50, 55, 60, 65, 70, 75, 80, 90, 58, 72, 85, 95],

    "Marks": [30, 35, 45, 50, 60, 68, 75, 90, 38, 65, 78, 95],

    "Sleep_Hours": [9, 8, 8, 7, 7, 6, 6, 5, 9, 7, 6, 5],

    "Assignments_Completed": [2, 3, 4, 5, 6, 7, 8, 10, 3, 7, 8, 11],

    "Performance": [
        "Poor", "Poor", "Average", "Average",
        "Average", "Good", "Good", "Excellent",
        "Poor", "Good", "Excellent", "Excellent"
    ]
}

df = pd.DataFrame(data)
````

## 1. Diagonal plots

You said the diagonal compares a feature with itself.

Technically, the diagonal displays the distribution of that feature.

Diagonal

Distribution of one feature.

Off-diagonal

Relationship between two features.


## 2. Default diagonal plot

You said KDE is the default. Small correction:
- Default kind = "scatter"
- Default diag_kind = "auto" — generally histogram without hue and KDE with hue.

## 3. kind and diag_kind

Remember this:
```markdown
Parameter	Controls	    Examples
kind	        Off-diagonal	    scatter, reg, kde, hist
diag_kind	Diagonal	    hist, kde
```

## 4. hue

Your gender example is exactly right.

For example, `hue="Gender"` assigns different colours to Male and Female observations. It doesn't filter or remove rows.

## Add ons

### 1. More study hours and higher marks

- Positive correlation means both variables tend to increase together.
- It does NOT prove that study hours alone cause higher marks.
- Other variables may influence the result.

### 2. Two performance categories are clearly separated

If two categories are clearly separated in a pairplot, it means the plotted features may help distinguish those categories.

For example:            
Imagine:             
- Poor-performing students appear mostly in one region.
- Excellent-performing students appear in another region.           
This suggests that the features might be useful for predicting student performance.

In Machine Learning, this is useful information for classification.

### 3. How can pairplot identify potential outliers?

Imagine most students have:         
- Study hours between 2 and 8.
- Marks between 30 and 90.
But one student has 15 study hours and 5 marks.         

That isolated point may be a potential outlier.

A pairplot helps because:               
- Off-diagonal scatterplots show unusual observations.
- Diagonal histograms or KDE plots can show unusually extreme values in individual features.
Remember: an outlier isn't automatically an error. We need to investigate it.           

### 4. No obvious linear relationship — are features useless?       

A feature can still be useful even if Pearson correlation is low.

For example, there could be a nonlinear relationship:

The relationship is clearly present, but a straight-line correlation measure may not capture it well.

Machine Learning models can sometimes learn nonlinear relationships.
