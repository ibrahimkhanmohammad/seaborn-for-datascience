## 1. What is a Pairplot?
Imagine you have a dataset containing 4 numerical columns:
- Study Hours
- Attendance
- Marks
- Sleep Hours
Previously, we used scatterplots to compare two numerical columns at a time.

For example:        
`sns.scatterplot(data=df, x="Hours_Studied", y="Marks")`

But what if you want to compare every numerical column against every other numerical column?

Creating individual scatterplots would take time.

That's where sns.pairplot() comes in.

A pairplot automatically creates multiple plots to show relationships between numerical variables in a dataset.

- Each non-diagonal plot shows the relationship between two numerical variables.
- Each diagonal plot shows the distribution of an individual variable.
In simple words:

Pairplot = Multiple scatterplots + individual distributions in one visualisation.

## 3. Understanding the Pairplot structure
This is the most important concept.

Diagonal = distribution of one variable


Other cells = relationships between two variables


Notice:
- The diagonal represents each variable against itself.
- The other cells represent relationships between different variables.
- The upper and lower triangles show the same pairwise relationships with reversed axes.
For 4 numerical features, we get a 4 × 4 matrix = 16 plots.

## 4. Understanding hue
Remember how we used hue in scatterplots and countplots?

It differentiates categories using different colours.

We can do the same in pairplots.
`sns.pairplot(    data=df,    hue="species")plt.show()`

What happens here?
- Setosa gets one colour.
- Versicolor gets another colour.
- Virginica gets another colour.
Now we can visually compare the distributions and relationships between different flower species.

For example, we might notice that petal length separates the species quite clearly.

Important: hue doesn't filter the dataset. It differentiates the observations according to their category.

## 5. Understanding kind
By default, pairplot uses scatterplots for non-diagonal cells.

But we can change the type of plot.         
### A. kind="scatter"
`sns.pairplot(    data=df,    kind="scatter")
`

Displays scatterplots between numerical variables.

This is the default.

### B. kind="reg"
`sns.pairplot(    data=df,    kind="reg")`


Displays scatterplots with regression lines.

Useful when exploring linear relationships.

### C. kind="kde"
`sns.pairplot(    data=df,    kind="kde")
`
Displays two-dimensional KDE plots in the off-diagonal cells.

It shows estimated density rather than individual scatter points.

### D. kind="hist"
`sns.pairplot(    data=df,    kind="hist")
`
Displays two-dimensional histograms in the off-diagonal cells.

Remember: kind controls the plots outside the diagonal.

## 6. Understanding diag_kind
What if you want to change the diagonal plots?

That's where diag_kind comes in.


### A. diag_kind="hist"
`sns.pairplot(    data=df,    diag_kind="hist")
`
Displays histograms on the diagonal.

### B. diag_kind="kde"
`sns.pairplot(    data=df,    diag_kind="kde")
`
Displays KDE curves on the diagonal.

Difference:
```markdown
    Parameter	    Controls
    kind	    Off-diagonal plots
    diag_kind	    Diagonal plots
```

This distinction is very important.

## 7. Understanding corner
Remember that the upper and lower triangles contain duplicate relationships.

We can display only one triangle using:
`sns.pairplot(    data=df,    corner=True)plt.show()`

What does it do?
- corner=False: displays both upper and lower triangles. Default.
- corner=True: displays only the lower triangle and diagonal.
This helps save space when working with many features.

## 8. Understanding vars and x_vars, y_vars
Sometimes you don't want every numerical column.

You only want to compare selected features.

A. vars
Suppose you only want to analyse:
- sepal_length
- petal_length
- petal_width           
`sns.pairplot(    data=df,    vars=["sepal_length", "petal_length", "petal_width"])plt.show()
`


vars specifies which numerical variables to include in the pairplot.

B. x_vars and y_vars
These give you more control over the comparison.            
`sns.pairplot(    data=df,    x_vars=["sepal_length", "petal_length"],    y_vars=["petal_width", "sepal_width"])plt.show()
`


Here:
- X-axis variables: sepal_length and petal_length.
- Y-axis variables: petal_width and sepal_width.
This creates a smaller grid containing only the selected comparisons.

## 9. Other useful parameters
```markdown
Parameter	            Purpose
height=2.5	            Height of each subplot in inches
aspect=1	            Aspect ratio of each subplot
palette="Set2"	            Controls colours used for hue categories
markers=["o", "s", "^"]	    Uses different marker shapes for hue categories
plot_kws={"alpha": 0.5}	    Passes settings to off-diagonal plots
diag_kws={"fill": True}	    Passes settings to diagonal plots
```

Example:
```python
sns.pairplot(    data=df,    hue="species",    palette="Set2",    height=2.5,    aspect=1,    markers=["o", "s", "^"],    plot_kws={"alpha": 0.6})
plt.show()

```


Note: The marker list must match the number of hue categories.

## 10. Why do we use Pairplots in Data Science?
Pairplots are particularly useful during Exploratory Data Analysis (EDA).

They help us:
1. Identify relationships between numerical features.
2. Discover possible positive or negative trends.
3. Compare distributions.
4. Identify clusters and separation between categories.
5. Spot potential outliers.
6. Explore which features might be useful for Machine Learning.
For example, in the Iris dataset, petal measurements help distinguish the three species.

But remember: a pairplot is an exploratory tool. It doesn't prove causation or tell us which features a model will definitely need.
