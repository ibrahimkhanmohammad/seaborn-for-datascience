## Task - an important EDA concept
1. How many different colors do you see?           
Three different colors
The Iris dataset contains three species:
- Setosa
- Versicolor
- Virginica

Each species gets a different color.


2. What does the legend represent?          
The legend explains which color corresponds to which species.

For example:            
- Blue → Setosa
- Orange → Versicolor
- Green → Virginica         
(The exact colors depend on the palette.)


3. What happens when you remove hue="species"?          
All observations are displayed using the same color, so we lose the visual distinction between species. 

Important: Matplotlib vs Seaborn
```console
    Matplotlib	                          Seaborn
    color="green"	                color="green"
    c=colors	                        hue="species"
    Manually assign colors to groups	Automatically assign colors based on categories
```

Remember these two parameters:
- color → One color for the entire plot.
- hue → Different colors based on a categorical variable.
