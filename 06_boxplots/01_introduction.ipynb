## 1. What is a box plot?
A box plot is used to visualize:
- Median
- Quartiles
- Data spread
- Potential outliers
Let's understand its structure.

## 2. Understand the five important components
Component	Meaning 
- Q1	First quartile (25th percentile)
- Q2	Median (50th percentile)
- Q3	Third quartile (75th percentile)        
IQR	Q3 − Q1     
Whiskers	Extend to the most extreme non-outlier observations
What are outliers?

Outliers are observations unusually far from the rest of the data.

A common box plot rule identifies potential outliers as:

$x < Q1 - 1.5(IQR)
$

$x > Q3 + 1.5(IQR)$
