Create a box plot with:
1. X-axis: Department
2. Y-axis: Salary
3. Figure size: (10, 6)
4. Color: skyblue
5. Title: Salary Distribution by Department
6. Appropriate axis labels.
7. Use fig, ax = plt.subplots() and sns.boxplot().

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {
"Department": [
"CSE", "CSE", "CSE", "CSE", "CSE",
"ECE", "ECE", "ECE", "ECE", "ECE",
"IT", "IT", "IT", "IT", "IT"
],
"Salary": [
25000, 28000, 30000, 32000, 80000,
22000, 25000, 27000, 29000, 35000,
30000, 32000, 35000, 38000, 90000
]
}

df = pd.DataFrame(data)
````

````python
fig, ax = plt.subplots(figsize=(10, 6))

sns.boxplot(
    data=df,
    ax=ax,
    x="Department",
    y="Salary",
    color="skyblue",
)
ax.set_title("Salary Distribution by Department")
plt.show()
````

## 2. Conceptual Quetsions
Q1. What does the line inside the box represent?            
It represents the median, also called Q2 or the 50th percentile.

Q2. What does IQR mean?                
Interquartile Range (IQR) is the difference between the third and first quartiles.

$IQ3 = Q3-Q1$


It represents the spread of the middle 50% of the data.

Q3. Why are some points outside the whiskers?           
Those are potential outliers.           
An outlier isn't necessarily an error.

It could be:
- A data entry mistake.
- An unusual observation.
- A genuine but extreme value.              
For example, a salary of ₹90,000 might be genuine even if most salaries are around ₹30,000.

Q4. Why is a box plot useful in Data Science?               
It helps us understand:
- Median.
- Quartiles.
- Data spread.
- Potential outliers.
- Differences between groups.
